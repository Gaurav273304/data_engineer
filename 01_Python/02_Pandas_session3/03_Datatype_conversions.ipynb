```python
import pandas as pd
df = pd.read_csv('employees.csv')
df.dtypes
```

```python

df = df.convert_dtypes()
df.dtypes
```

```python
df['join_date'] = pd.to_datetime(df['join_date'])
df.dtypes
```

```python
#the salary column has -999 sentinal values.Replace them With NaN 
#and then convert salary column to numeric data type

df['salary'] = pd.to_numeric(df['salary'].replace(-999,pd.NA))
df.dtypes
```
