```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('dirty_cafe_sales.csv')
df.info()
```

```python
#2find total number of missing values in entire dataframe and in each column

df.isnull().sum().sum()
#3.for each column percentage of missing values -> loction has highest
missing_percentage = df.isnull().sum()/len(df)*100
print(missing_percentage)
```

```python
#4. Display all rows which have at least one missing value
missing_values = df[df.isnull().any(axis=1)]
print(missing_values)

```

```python
#5 find duplicates rows
duplicate_rows = df[df.duplicated()]
print(duplicate_rows)
```

```python
col = ['Quantity','Price Per Unit','Total Spent']
#replace multiple values with NaN
df[col] = df[col].replace(['ERROR','UNKNOWN'], np.nan)
df 
```

```python
df['Quantity'] = pd.to_numeric(df['Quantity'],errors='coerce')
df['Total Spent'] = pd.to_numeric(df['Total Spent'],errors='coerce')
df
```

```python
df['Transaction Date'] = pd.to_datetime(df['Transaction Date'],errors = 'coerce')
df.dtypes
```

```python

convert_dict = {'Item': 'category','Payment Method': 'category','Location': 'category'}
df = df.astype(convert_dict)
df.info(memory_usage= 'deep')
```

```python
df['Payment Method'] = df['Payment Method'].fillna(df['Payment Method'].mode()[0])  #fill with most frequent value
#fill the location with most frequent value
df['Location'] = df['Location'].fillna(df['Location'].mode()[0])
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].median())  #fill with median value
df.isnull().sum()
```

```python
df['Item'] = df['Item'].str.strip().str.title()
df['Item'] = df['Item'].fillna(df['Item'].mode()[0])
df.head()
```

```python
df.isnull().sum()
```

```python
#drop all missing values
df.dropna(inplace = True)
df.isnull().sum()
```

```python
rename_columns = {'Transaction ID': 'txn_id','Transaction Date':'txn_date','Price Per Unit':'price_per_unit','Total Spent':'total_spent','Payment Method':'payment_method'}
df.rename(columns = rename_columns,inplace = True)
# 
df.head()
```

```python
#make sure location and Quantity column name are in lower case
df.rename(columns = {'Location':'location','Quantity':'quantity'},inplace = True)
df.head() 
```
