```python
import pandas as pd
import numpy as np
df = pd.read_csv("employees.csv")
df['salary']
```

```python
# apply a custom function
def select_category(salary):
    if salary >= 120000:
        return 'High'
    elif salary >= 80000:
        return 'Medium'
    else:
        return 'Low'
    
def summary(row):
    return f"{row['name']} has a salary of {row['salary']}"


df['salary_category'] = df['salary'].apply(select_category)
df['summary'] = df.apply(summary, axis=1)

df['last_name'] = df['name'].apply(lambda name: name.split()[-1])
df[["salary_category", "salary", "name", "last_name", "summary"]]
```

```python
df['bonus_elegibile'] = df['rating'].apply(lambda rating: 'Yes' if rating >= 4 else 'No')
df[['rating','bonus_elegibile']]
```

```python
# df['no_of_days_in_company']
df['join_date'] = pd.to_datetime(df['join_date'],format = 'mixed')
# df['days_since_joining'] = (pd.Timestamp.now() - df['join_date']).dt.days

df['days_since_joining'] = df['join_date'].apply(lambda date: (pd.Timestamp.now() - date).days)
df['years_since_joining'] = df['days_since_joining'] // 365
df[['join_date','days_since_joining','years_since_joining']]
```
