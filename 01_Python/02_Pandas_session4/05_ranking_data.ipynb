```python
import pandas as pd
import numpy as np
df = pd.read_csv("employees.csv")
df['salary'] = df['salary'].fillna(0).astype(int)
df['salary_rank'] = df['salary'].rank(ascending=False) #default is ascending
new_df = df.sort_values('salary_rank')
new_df[['emp_id', 'salary', 'salary_rank']]
```

```python
#using dense -> dense use the same rank for the same value and does not skip the next rank.
df['salary_rank_dense'] = df['salary'].rank(method='dense', ascending=False)
new_df = df.sort_values('salary_rank_dense')
new_df[['emp_id', 'salary', 'salary_rank_dense', 'salary_rank']]
```

```python
# method = 'first' -> assigns ranks in the order they appear in the array. This is the default method.
# method = 'min' -> assigns the minimum rank to all the tied values.
# method = 'max' -> assigns the maximum rank to all the tied values.
# method = 'dense' -> assigns ranks in the order they appear in the array, but skips the next rank if the value is the same.    
```

```python
#Engineering department me salaey basics par rank karna hai

df['dept_salary_rank'] = df.groupby('department')['salary'].rank(ascending=False,method='dense')
df[['emp_id','salary','department','dept_salary_rank']].sort_values(['department','dept_salary_rank'])
```
