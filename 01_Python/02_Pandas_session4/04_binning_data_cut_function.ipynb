```python
import pandas as pd
import numpy as np
df = pd.read_csv("employees.csv")
df['salary'] =df['salary'].fillna(0).astype(int)

#bins are (left, right] meaning that the left edge is exclusive and the right edge is inclusive.
#(0,40000] ->  greater than 0 and <= 40000 

bins = [0,40000,70000,100000,150000,300000]
labels = ['Very Low', 'Low', 'Medium', 'High', 'Very High']

df['salary_level']= pd.cut(df['salary'], bins=bins, labels=labels,right = False)
# if right = False then the bins are [left, right) meaning that the left edge is inclusive and the right edge is exclusive.
df[['salary','salary_level']]
```

```python
# using qcut
'''
low->0
-10000
-20000
-30000
highest -> 400000

-> means equal intervals

q1,q2,q3,q4

'''

salary_quartiles = pd.qcut(df['salary'],q=4,labels=['Q1','Q2','Q3','Q4'])
df['salary_quartiles'] = salary_quartiles
df[['salary','salary_level', 'salary_quartiles']]
```
