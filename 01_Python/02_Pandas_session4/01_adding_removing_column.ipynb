```python
import pandas as pd
```

```python
df = pd.read_csv("employees.csv")
df.isnull().sum()
```

```python
df['salary'] = df['salary'].fillna(0).astype(int)
df['annual_salary'] = df['salary'] * 12
df[['salary','annual_salary']]
df['tax'] = df['annual_salary'] * 0.3
df['inhand_salary'] = df['annual_salary'] - df['tax']
df[['annual_salary','tax','inhand_salary']]
```

```python
import numpy as np
df['is_senior'] = np.where(df['experience_years']>=20,'Yes','No')
conditions = [
    df['salary'] < 100000,
    (df['salary'] >= 100000) & (df['salary'] < 250000),
    df['salary'] >= 250000
]
choices = ['Low','Medium','High']
df['salary_level'] = np.select(conditions,choices,default='Unknown')
df[['salary','salary_level','is_senior']]
```
